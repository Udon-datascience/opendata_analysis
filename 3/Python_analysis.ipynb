```python
'''
分析の指針を整理する。

＜目的＞
データの分析により、新たな知見を見つけ出し、商品の売り上げを向上させること。


＜分析内容＞
・商品カテゴリーごとの分析
・支払方法ごとの変化
・利益が最も大きいカテゴリー
・顧客分析、ペルソナの考察
・時期ごとの注文カテゴリの傾向分析

したがって、下記のような分析が候補として考えられる。
・売上に影響している要因は何か（回帰）
・利益を出す商品・出さない商品を予測する（分類）
・ペルソナの推定（クラスタリング、分類）

今回は回帰分析の練習として、「売上に影響している要因は何か（回帰）」に絞って分析を行う。

今回のデータセットでは、mv(市場価値)は明らかにされているものの、原価や経費が明らかにされていない。したがって、今回は利益ではなく売上を目的変数とする。
売上を目的変数にするためには、'price', 'discount_amount', 'qty_ordered', 'grand_total'などから売上に相当する列を作る必要がある。
したがってこれらの列を適切に説明変数か目的変数に設定する必要がある。
そのために、以下では不明な'grand_total'の定義を明らかにする。
'''
```

```python
cols = ['price', 'discount_amount', 'qty_ordered', 'grand_total']
df2 = df[cols]
df2.head(20)
#このセルの結果より、grand_total = (price - discount_amount) *qty_ordered 
#もしくは grand_total = price * qty_ordered - discount_amount と考えられる。
```

```python
cheker = df2[(df2['discount_amount'] > 0) & (df2['qty_ordered'] > 2)]
display(cheker)
#このセルの結果より、 grand_total = price * qty_ordered - discount_amount と考えられる。
```
