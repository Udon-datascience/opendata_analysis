```python
query = """
-- データ件数を確認
SELECT COUNT(*)
FROM ecommerce_orders;
"""
df = pd.read_sql(query, engine)
display(df)
```

```python
query = """
-- データを確認
SELECT *
FROM ecommerce_orders
LIMIT 10;
"""
df = pd.read_sql(query, engine)
display(df)
```
